```python
%pip install pandas openpyxl xlrd
from pathlib import Path
import pandas as pd

folder = Path.cwd()
print(f"Searching in: {folder}\n")

# Find data files in this folder and all its subfolders.
files = sorted(
    path for path in folder.rglob("*")
    if path.is_file()
    and path.suffix.lower() in [".xlsx", ".xls", ".csv"]
    and not path.name.startswith(("~$", "._"))
)

successful = 0

for path in files:
    # Show the subfolder path as well as the filename.
    relative_path = path.relative_to(folder)

    try:
        if path.suffix.lower() == ".csv":
            df = pd.read_csv(path, nrows=5)
        else:
            df = pd.read_excel(path, nrows=5)

        print(f"SUCCESS: {relative_path}")
        print(f"  First five columns: {list(df.columns[:5])}\n")
        successful += 1

    except Exception as error:
        print(f"ERROR: {relative_path}\n  {error}\n")

print(f"Finished: {successful}/{len(files)} files read successfully.")
```

The following tests one method of writing new files to the cleaned folder, for organization sake 

```python
from pathlib import Path
import pandas as pd

folder = Path.cwd()

# Load the full file, rather than the five-row test sample.
countries = pd.read_excel(folder / "Country list.xlsx")

# Make a copy and apply basic cleaning.
countries_cleaned = countries.copy()
countries_cleaned.columns = countries_cleaned.columns.str.strip()

for column in countries_cleaned.select_dtypes(include="object").columns:
    countries_cleaned[column] = countries_cleaned[column].apply(
        lambda value: value.strip() if isinstance(value, str) else value
    )

# Ensure the output folder exists.
cleaned_folder = folder / "Cleaned"
cleaned_folder.mkdir(exist_ok=True)

# Save the cleaned copy without an extra index column.
output_path = cleaned_folder / "Country_list_cleaned.xlsx"
countries_cleaned.to_excel(output_path, index=False)

print(f"Saved: {output_path}")
print(f"Rows: {len(countries_cleaned):,}")
display(countries_cleaned.head())
```
